# ACTIVATION FUNCTIONS- TODO: IMPLEMENT

```python
def relu(z):
    return np.maximum(0, z)

def relu_derivative(z):
    return (z > 0).astype(float)
```

# LOSS FUNCTION- TODO: IMPLEMENT

```python
def mse_loss(y_true, y_pred):
  """Calculates the Mean Squared Error between true and predicted values."""
  # The formula is: sum((y_true - y_pred)^2) / number of samples
  error = y_true - y_pred
  return np.mean(error**2)
```

## **Xavier (Glorot) Initialization**

When training neural networks, how we initialize weights matters.  
- If weights are **too small** → activations and gradients vanish.  
- If weights are **too large** → activations and gradients explode.  

#**Xavier initialization** (Glorot & Bengio, 2010) balances this by keeping the variance of activations roughly the same across all layers.

### In This Assignment
- **W1 (input → hidden1):** `fan_in = input_dim`, `fan_out = hidden1`  
- **W2 (hidden1 → hidden2):** `fan_in = hidden1`, `fan_out = hidden2`  
- **W3 (hidden2 → output):** `fan_in = hidden2`, `fan_out = output_dim`  

Your task: compute the correct `xavier_std` for each layer, sample weights from a normal distribution with that std, and set biases = 0.



```python
def xavier_initialization(input_dim, hidden1, hidden2, output_dim):
    """
    Initializes network weights using the Xavier (Glorot) method.
    
    This method scales weights to keep the variance of activations consistent 
    across layers, preventing vanishing/exploding gradients.
    """
    np.random.seed(assignment["student_seed"])

    # Calculate the scale factor for each layer's weights
    scale_w1 = np.sqrt(2.0 / (input_dim + hidden1))
    scale_w2 = np.sqrt(2.0 / (hidden1 + hidden2))
    scale_w3 = np.sqrt(2.0 / (hidden2 + output_dim))

    # Initialize weights from a normal distribution and scale them
    W1 = np.random.randn(input_dim, hidden1) * scale_w1
    W2 = np.random.randn(hidden1, hidden2) * scale_w2
    W3 = np.random.randn(hidden2, output_dim) * scale_w3

    # Initialize biases to zeros
    b1 = np.zeros((1, hidden1))
    b2 = np.zeros((1, hidden2))
    b3 = np.zeros((1, output_dim))

    return W1, b1, W2, b2, W3, b3
```

# FORWARD PROPAGATION - TODO: IMPLEMENT

```python
def forward_pass(X, W1, b1, W2, b2, W3, b3):
    """
    Performs the forward pass to compute predictions.
    
    The data flows through the network:
    Input -> Hidden1(relu) -> Hidden2(relu) -> Output(Linear)
    """
    # Compute the pre-activation and activation for the first hidden layer
    z_layer1 = np.dot(X, W1) + b1
    layer1_output = relu(z_layer1)
    
    # Compute the pre-activation and activation for the second hidden layer
    z_layer2 = np.dot(layer1_output, W2) + b2
    layer2_output = relu(z_layer2)
    
    # Compute the final output (pre-activation, no activation function here)
    z_output = np.dot(layer2_output, W3) + b3
    
    # The pre-activations are useful for backpropagation
    return z_layer1, layer1_output, z_layer2, layer2_output, z_output
```

# BACKWARD PROPAGATION - TODO:IMPLEMENT

```python
def backward_pass(X, Y_true, z1, a1, z2, a2, Y_pred, W2, W3):
    """
    Performs backpropagation to calculate gradients for all parameters.
    
    Gradients are calculated using the chain rule, propagating the error
    backward from the output layer to the input layer.
    """
    num_samples = len(X)
    
    # Start with the gradient of the loss with respect to the output prediction
    grad_output = 2 * (Y_pred - Y_true) / num_samples
    
    # Compute gradients for the output layer (W3, b3)
    dW3 = np.dot(a2.T, grad_output)
    db3 = np.sum(grad_output, axis=0, keepdims=True)

    # Propagate gradient back to the second hidden layer
    grad_a2 = np.dot(grad_output, W3.T)
    grad_z2 = grad_a2 * relu_derivative(z2)
    
    # Compute gradients for the second hidden layer (W2, b2)
    dW2 = np.dot(a1.T, grad_z2)
    db2 = np.sum(grad_z2, axis=0, keepdims=True)
    
    # Propagate gradient back to the first hidden layer
    grad_a1 = np.dot(grad_z2, W2.T)
    grad_z1 = grad_a1 * relu_derivative(z1)

    # Compute gradients for the first hidden layer (W1, b1)
    dW1 = np.dot(X.T, grad_z1)
    db1 = np.sum(grad_z1, axis=0, keepdims=True)

    return dW1, db1, dW2, db2, dW3, db3
```

# TRAINING FUNCTION - TODO:  COMPLETE IMPLEMENTATION

```python
def train_neural_network(X_train, Y_train, X_test, Y_test, epochs=500, patience=10):
    """
    Trains the neural network using forward and backward propagation.
    """
    # Initialize weights and biases
    W1, b1, W2, b2, W3, b3 = xavier_initialization(1, hidden1, hidden2, 1)

    best_test_loss = float('inf')
    best_weights = None
    patience_counter = 0

    train_losses = []
    test_losses = []

    print("Starting training...")
    print(f"Architecture: 1 -> {hidden1} -> {hidden2} -> 1")
    print(f"Learning Rate: {learning_rate}")
    print(f"Max Epochs: {epochs}, Early Stopping Patience: {patience}")
    print("-" * 50)

    for epoch in range(epochs):
        # Forward pass and training loss calculation
        z1, a1, z2, a2, y_pred_train = forward_pass(X_train, W1, b1, W2, b2, W3, b3)
        train_loss = mse_loss(Y_train, y_pred_train)

        # Backward pass to get gradients
        dW1, db1, dW2, db2, dW3, db3 = backward_pass(X_train, Y_train, z1, a1, z2, a2, y_pred_train, W2, W3)

        # Update parameters using gradient descent
        W1 -= learning_rate * dW1
        b1 -= learning_rate * db1
        W2 -= learning_rate * dW2
        b2 -= learning_rate * db2
        W3 -= learning_rate * dW3
        b3 -= learning_rate * db3

        # Evaluate performance on the test set
        _, _, _, _, y_pred_test = forward_pass(X_test, W1, b1, W2, b2, W3, b3)
        test_loss = mse_loss(Y_test, y_pred_test)

        train_losses.append(train_loss)
        test_losses.append(test_loss)

        if (epoch + 1) % 20 == 0:
            print(f"Epoch {epoch+1:3d}: Train Loss = {train_loss:.6f}, Test Loss = {test_loss:.6f}")

        # Early stopping logic
        if test_loss < best_test_loss:
            best_test_loss = test_loss
            best_weights = (W1.copy(), b1.copy(), W2.copy(), b2.copy(), W3.copy(), b3.copy())
            patience_counter = 0
        else:
            patience_counter += 1

        if patience_counter >= patience:
            print(f"Early stopping triggered at epoch {epoch+1}")
            print(f"Best test loss: {best_test_loss:.6f}")
            break

    return best_weights, train_losses, test_losses
```

# EXECUTE TRAINING

```python
print("Training Neural Network with your specific configuration...")
weights, train_losses, test_losses = train_neural_network(
    X_train_scaled, Y_train_scaled, X_test_scaled, Y_test_scaled,
    epochs=500, patience=10
)
```
